```python
import re
from collections import defaultdict
from datetime import datetime
import math

fileIn = open('dataset_.txt', 'r')
fileOut = open('out.txt', 'w')
```

```python
listToStr=""
ls2grm = []
n=2
regex = re.compile('[\w\-\']+')
start_time = datetime.now()
for line in fileIn :  
    matches = regex.findall(line)
    if matches: 
        grams = [' '.join(matches[i:i+n]) for i in range(len(matches)-n)]
        for gram in grams:
            ls2grm.append(gram)
frq = {}
for elem in ls2grm:
    if elem in frq:
        value = frq[elem]
        frq[elem]=value+1
    else:
        frq[elem]=1
        
fileOut.write("%s\n" % frq)
end_time = datetime.now()
print(f"Duration: {end_time - start_time}")
#prnt (frq)

fileOut.close()
```

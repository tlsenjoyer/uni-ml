```python
from itertools import islice

def printFirstN(d: dict, n: int):
    for key, value in islice(d.items(), n):
        print(key, value)
```

```python
import re
from collections import defaultdict
from datetime import datetime
import math

fileIn = open('dataset_.txt', 'r')
fileOut = open('out.txt', 'w')
```

```python
listToStr=""
grams = []
n = 2
regex = re.compile('[\w\-\']+')


for line in fileIn :  
    matches = regex.findall(line)
    if matches: 
        lineGrams = [' '.join(matches[i:i+n]) for i in range(len(matches)-n+1)]
        for gram in lineGrams:
            grams.append(gram)

gramOccurrences = defaultdict(int)
for gram in grams:
    if gram:
        gramOccurrences[gram] += 1
        
fileOut.write("%s\n" % gramOccurrences)

printFirstN(gramOccurrences, 10)
```

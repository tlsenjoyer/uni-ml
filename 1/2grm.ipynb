```python
import sys
import re
import codecs
from datetime import datetime

fin = codecs.open('c:/your_path/dataset_.txt', 'r', 'utf8')
fout = codecs.open('c:/your_path/2grm.txt', 'w', 'utf8')
```

```python
listToStr=""
ls2grm = []
n=2
regex = re.compile('[\w\-\']+')
start_time = datetime.now()
for line in fin :  
    result = regex.findall(line)
    if result: 
        grams = [' '.join(result[i:i+n]) for i in range(len(result)-n)]
        for gram in grams:
            ls2grm.append(gram)
frq = {}
for elem in ls2grm:
    if elem in frq:
        value = frq[elem]
        frq[elem]=value+1
    else:
        frq[elem]=1
        
fout.write("%s\n" % frq)
end_time = datetime.now()
print('Duration: {}'.format(end_time - start_time))
#prnt (frq)
```

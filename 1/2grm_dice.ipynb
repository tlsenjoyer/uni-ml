```python
def printFirstFew(s: dict):
    i=0
    for key in s:
        if i < 10:
            print (key)
            i=i+1
        else:
            break
```

```python
import re
from collections import defaultdict
import math

fileIn = open('dataset_.txt', 'r')
fileOut = open('out_dice.txt', 'w')
```

```python
grams = []
wordOccurences = defaultdict(int)
gramOccurences = defaultdict(int)
diceCoeffs = defaultdict(float)
n=2
regex = re.compile('[\w\-\']+')
for line in fileIn:  
    result = regex.findall(line)
    if result: 
        lineGrams = [' '.join(result[i:i+n]) for i in range(len(result)-n+1)] # тут была ошибка, не учитывалось последнее слово
        for gram in lineGrams:
            grams.append(gram)

        for lineWord in result:
            if lineWord :
                wordOccurences[lineWord] += 1


for gram in grams:
    if gram:
        gramOccurences[gram]+=1
```

```python
def dice_coefficient(a, b, c):
    """dice coefficient 2c/(a + b)."""
    return  c * 2.0/(a + b) 
```

```python
def dice_log_coefficient(a, b, c):
    """dice coefficient 2c/(a + b)."""
    return -1 * math.log( c * 2.0/(a + b) )
```

```python
for gram in gramOccurences:
    gramWords = gram.split()
    diceCoeffs[gram] = dice_log_coefficient(wordOccurences[gramWords[0]],wordOccurences[gramWords[1]],gramOccurences[gram]) 

```

```python
printFirstFew(diceCoeffs)
```

```python
sortedDiceCoeffs = sorted(diceCoeffs, key=lambda x: diceCoeffs[x], reverse=True)
```

```python
for gram in sortedDiceCoeffs:
    s = f"{gram}\t{diceCoeffs[gram]}\n"
    fileOut.write(s)
print('ok')
fileOut.close()
```

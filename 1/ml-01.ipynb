```python
import sys
import re
import codecs

fin = codecs.open('dataset_.txt', 'r', 'utf8')
fout = codecs.open('frq.txt', 'w', 'utf8')
                
```

```python
from collections import defaultdict

regex = re.compile('[\w\-\']+')
lsWord = defaultdict(int)
for line in fin : 
    result = regex.findall(line)
    if result:         
        for elem in result:
            if not elem :
                continue
            lsWord[elem] += 1
```

```python
sorted_keys = sorted(lsWord, key=lambda x: int(lsWord[x]), reverse=True)
```

```python
for key in sorted_keys:
    s = str('{0}\t{1}\n').format(key,lsWord[key])
    fout.write(s)
print(sorted_keys[:10])
#print('ok!')
fout.close()
```

```python
x=[]
y=[]
i=0
for key in sorted_keys:
    y.append(lsWord[key])
    i=i+1
    x.append(i)
```

```python
%matplotlib notebook
import numpy as np
import matplotlib.pyplot as plt


# Построение графика
plt.title("частотное распределение") # заголовок
plt.xlabel("слова") # ось абсцисс
plt.ylabel("частота") # ось ординат
plt.grid() # включение отображение сетки
#plt.xlim(0, 100)  # Ограничение оси x 

plt.plot(x,y)
plt.show()
 
```

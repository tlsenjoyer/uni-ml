# PMI (pointwise mutual information)

Following `ML-01.pdf`, page 5:

$$PMI(A, B) = \log_2 \frac{P(A, B)}{P(A)P(B)} = \log_2 \frac{cN}{ab}$$

Here `a` and `b` are word counts, `c` is the bigram count, and `N` is the total number of words. We estimate `P(A) = a/N`, `P(B) = b/N`, and `P(A, B) = c/N`. As in `2grm_dice.ipynb`, co-occurrence means an ordered pair of adjacent words within one line. Only observed bigrams are scored. Higher PMI means stronger association; negative scores are preserved.

Run the notebook from the `1` directory. Results are written to `out_pmi.txt` in descending PMI order.

```python
def printFirstFew(s: dict):
    i=0
    for key in s:
        if i < 10:
            print(key)
            i=i+1
        else:
            break
```

```python
import re
from collections import defaultdict
import math
```

```python
grams = []
wordOccurences = defaultdict(int)
gramOccurences = defaultdict(int)
pmiCoeffs = defaultdict(float)
n = 2
regex = re.compile(r"[\w\-']+")

with open('dataset_.txt', 'r', encoding='utf-8') as fileIn:
    for line in fileIn:
        matches = regex.findall(line)
        if matches:
            lineGrams = [' '.join(matches[i:i+n]) for i in range(len(matches)-n+1)]
            for gram in lineGrams:
                gramOccurences[gram] += 1

            for lineWord in matches:
                wordOccurences[lineWord] += 1

totalWords = sum(wordOccurences.values())
```

```python
def pmi_coefficient(a, b, c, total_words):
    """PMI in bits: log2(c * total_words / (a * b)), for positive counts."""
    return math.log2(c * total_words / (a * b))
```

```python
for gram in gramOccurences:
    gramWords = gram.split()
    pmiCoeffs[gram] = pmi_coefficient(
        wordOccurences[gramWords[0]],
        wordOccurences[gramWords[1]],
        gramOccurences[gram],
        totalWords
    )
```

```python
printFirstFew(pmiCoeffs)
```

```python
sortedPmiCoeffs = sorted(pmiCoeffs, key=lambda x: pmiCoeffs[x], reverse=True)
```

```python
with open('out_pmi.txt', 'w', encoding='utf-8') as fileOut:
    for gram in sortedPmiCoeffs:
        s = f"{gram}\t{pmiCoeffs[gram]}\n"
        fileOut.write(s)
print('ok')
```
